# RestAPI **Methods**

```python
# GET

import requests

url="https://api.restful-api.dev/objects/ff808181a09d98f701a10069350868c2"

get= requests.get(url)

data=get.json()
print(f"Data: {data}")
print(get.status_code)
```

```python
# POST

import requests
import json

url="https://api.restful-api.dev/objects/"

payload= {
  "name": "Samsung Galaxy 27 Ultra",
  "data": {
    "year": 2027,
    "price": 1849.99,
    "CPU model": "Snapdragon 8 extreme gen 6 for Galaxy ",
    "Hard disk size": "1 TB"
  }
}

response= requests.post(url, json=payload ,timeout=10)
data = response.json()
print(f"Data: {data}")
print(response.status_code)
print(response.headers)
print(response.headers.get("content-type"))


```

```python
# PUT

import requests
import json

object_id= "ff808181a09d98f701a10069350868c2"

url=f"https://api.restful-api.dev/objects/{object_id}"

payload = {
    "name": "Samsung Galaxy 27 Ultra Updated",
    "data": {
        "year": 2027,
        "price": 1999.99,
        "CPU model": "Snapdragon 8 Elite",
        "Hard disk size": "1 TB"
    }
}

response= requests.put(url, json=payload ,timeout=10)
data = response.json()
print(f"Data: {data}")
print(response.status_code)


```

```python
# PATCH

import requests
import json

object_id= "ff808181a09d98f701a10069350868c2"

url=f"https://api.restful-api.dev/objects/{object_id}"

payload = {
    "name": "Samsung Galaxy 27 Ultra Updated",
    "data":{
       "price": 1899.99
    }

}

response= requests.patch(url, json=payload ,timeout=10)
data = response.json()
print(f"Data: {data}")
print(response.status_code)



```

```python
# DELETE

import requests
import json

object_id= "ff808181a09d98f701a10069350868c2"

url=f"https://api.restful-api.dev/objects/{object_id}"

payload = {
    "name": "Samsung Galaxy 27 Ultra Updated",
    "data":{
       "price": 1899.99
    }

}

response= requests.delete(url, json=payload ,timeout=10)
data = response.json()
print(f"Data: {data}")
print(response.status_code)


```
